```python
from retrieval_updated import (
    load_or_build_chunk_search_index,
    load_or_build_transcript_search_index,
)

chunk_index = load_or_build_chunk_search_index()
transcript_index = load_or_build_transcript_search_index()

print("Loaded chunk index keys:")
print(chunk_index.keys())
print()
print("Loaded transcript index keys:")
print(transcript_index.keys())
```

```python
def show_dimension(index, dim):
    terms = index["dimension_terms"][dim]
    print(f"Dimension {dim}")
    print("Positive terms:")
    print(", ".join(terms["positive"]))
    print()
    print("Negative terms:")
    print(", ".join(terms["negative"]))
    print("-" * 80)
```

```python
print("Number of chunk dimensions:", len(chunk_index["dimension_terms"]))
print("Number of transcript dimensions:", len(transcript_index["dimension_terms"]))
```

```python
# Show the first 10 chunk dimensions
for dim in range(min(10, len(chunk_index["dimension_terms"]))):
    show_dimension(chunk_index, dim)
```

```python
# Show the first 10 transcript dimensions
for dim in range(min(10, len(transcript_index["dimension_terms"]))):
    show_dimension(transcript_index, dim)
```

```python
# Inspect a specific dimension manually
show_dimension(chunk_index, 0)
show_dimension(chunk_index, 1)
show_dimension(chunk_index, 2)
```

```python
# Compare the same dimension across chunk and transcript indexes
dim = 0

print("CHUNK INDEX")
show_dimension(chunk_index, dim)

print("TRANSCRIPT INDEX")
show_dimension(transcript_index, dim)
```

```python
import pandas as pd

pd.set_option('display.max_rows', None)       # show all rows
pd.set_option('display.max_columns', None)    # show all columns
pd.set_option('display.max_colwidth', None)   # show full text in cells
pd.set_option('display.width', None)          # prevent line wrapping cutoff

def dimensions_to_df(index, n=100):
    rows = []
    for dim in range(min(n, len(index["dimension_terms"]))):
        terms = index["dimension_terms"][dim]
        rows.append({
            "dimension": dim,
            "positive_terms": ", ".join(terms["positive"]),
            "negative_terms": ", ".join(terms["negative"]),
        })
    return pd.DataFrame(rows)

dimensions_to_df(transcript_index, 100)
```
